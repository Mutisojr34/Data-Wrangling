```python
# Display some values from the Location column
fatality_locations["Location"].head(10)
```

```python
# Check for missing values in the Location column
print("Missing Location values:", fatality_locations["Location"].isnull().sum())
```

```python
# Test splitting the Location column at commas
fatality_locations["Location"].dropna().head(10).str.split(",")
```

```python
# Split Location into Region and State/Country
fatality_locations[["Region", "State/Country"]] = (
    fatality_locations["Location"].str.split(",", n=1, expand=True)
)

fatality_locations.head(10)
```

```python
# Check the column names after splitting Location
print(fatality_locations.columns)
```

```python
# Check the Region and State/Country columns
fatality_locations[["Location", "Region", "State/Country"]].head(10)
```

```python
# Sort the dataframe by fatalities from highest to lowest
fatality_locations_sorted = fatality_locations.sort_values(
    by="Fatalities",
    ascending=False
)

fatality_locations_sorted.head(10)
```

```python
# Select the first 100 crashes with the highest fatalities
top_100_fatality_crashes = fatality_locations_sorted.head(100)

top_100_fatality_crashes
```

```python
# Check the first and last fatality values in the top 100
print("Highest fatalities:", top_100_fatality_crashes["Fatalities"].iloc[0])
print("Lowest fatalities in top 100:", top_100_fatality_crashes["Fatalities"].iloc[-1])
```

```python
top_100_fatality_crashes
```

```python
# Display the State/Country values from the top 100 crashes
top_100_fatality_crashes["State/Country"].head(20)
```

```python
top_25_fatality_crashes = fatality_locations_sorted.head(25)

top_25_fatality_crashes[["State/Country", "Fatalities"]]
```

```python
top_25_country_fatalities = (
    top_25_fatality_crashes.groupby("State/Country")["Fatalities"]
    .sum()
    .sort_values(ascending=False)
)

top_25_country_fatalities
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

colors = sns.color_palette("pastel", len(top_25_country_fatalities))

plt.figure(figsize=(10, 10))

plt.pie(
    top_25_country_fatalities,
    labels=top_25_country_fatalities.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)

plt.title("Distribution of Fatalities in Top 25 Crashes by State/Country")
plt.show()
```
